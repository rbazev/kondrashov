```python
from Bio import AlignIO
import re
from pathlib import Path

input_folder = Path(".")
output_folder = Path("renamed_alignments")
output_folder.mkdir(parents=True, exist_ok=True)

for infile in input_folder.iterdir():
     # Process only alignment files
    if not infile.is_file() or infile.suffix.lower() != ".aln":
        continue

    alignment = None
    input_format = None

    try:
        alignment = AlignIO.read(infile, "fasta")

        for record in alignment:
            match = re.search(r"\[([^\]]+)\]", record.description)

            if match:
                species = match.group(1).replace(" ", "_")
                record.id = species
                record.name = species
                record.description = species
            else:
                print(f"No species name found: {record.id} in {infile.name}")

        # Save with the original filename
        outfile = output_folder / f"{infile.stem}_renamed.fasta"

        AlignIO.write(alignment, outfile, "fasta")

        print(f"Processed: {infile.name} -> {outfile.name}")

    except Exception as e:
        print(f"Skipped {infile.name}: {e}")

print("Finished processing alignment files.")

```

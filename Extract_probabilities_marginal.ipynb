### use ff_dict to automatically retrieve the corresponding prob.marginal.txt file for each gene, then extract the lines matching each gene's specific ancestral node and amino acid position.

```python
import os
import csv

def read_marginal_files(ff_dict, directory= "C:\\Users\\adesi\\OneDrive - University Of Houston\\Documents\\Ricardo_lab\\iqtree-3.1.4-Windows\\Ancestral_State_Reconstruction\\ASR_Correct_files"):

    marginal_data = {}

    for protein, nodes in ff_dict.items():

        file_path = os.path.join(
            directory, protein, 'prob.marginal.txt'
        )

        if not os.path.isfile(file_path):
            print(f'File not found: {protein}')
            continue

        with open(file_path, 'r') as f:
            lines = f.readlines()

        marginal_data[protein] = {}

        # Iterate through each ancestral node
        for node, sites in nodes.items():

            marginal_data[protein][node] = {}

            # Extract the required amino acid positions
            for site in sites:

                search_string = f'{node},{site}'

                matching_lines = [
                    line.strip().replace(',', '\t')
                    for line in lines
                    if line.startswith(search_string + ',')
                ]

                marginal_data[protein][node][site] = matching_lines

        print(f'Processed: {protein}')

    return marginal_data


marginal_data = read_marginal_files(ff_dict)
```

```python
with open('marginal_data_compiled.csv', 'w', newline='') as f:

    writer = csv.writer(f, delimiter=',')

    writer.writerow([
        'GENE', 'Tree_node', 'site', 'Marginal_data'
    ])

    for protein, nodes in marginal_data.items():

        for node, sites in nodes.items():

            for site, matching_lines in sites.items():

                for line in matching_lines:
                    writer.writerow([
                        protein,
                        node,
                        site,
                        *line.split('\t')[2:]
                    ])

print('Saved as marginal_data.csv')
```

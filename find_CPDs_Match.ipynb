### Step 2

Take the fasta files containing ortholog sequences, search for human sequences and extract the corresponding record.

```python
import os
os.listdir('fasta_match')
```

### This code works for fasta_match/ directory which has a different naming scheme

```python
import os
from Bio import SeqIO

protrecs = {}
i = 1
for filename in os.listdir("fasta_match"):
    if not filename.endswith(".fasta"):
        continue
    gene = filename.split("_")[0]
    protrecs.setdefault(gene, [])
    filepath = os.path.join("fasta_match", filename)

    for record in SeqIO.parse(filepath, "fasta"):
        if get_species(record) == "Homo sapiens":
            protrecs[gene].append(record)
            print(i, gene, record.id, len(record))
            i += 1
```

### STEP 3 
##### Match variant transcript IDs with ortholog proteins

Take variant transcript ID, query the NCBI database, extract the corresponding protein, and look for it in the ortholog set.  Flag problems.

**Output:** `translation` dictionary with transcript IDs as keys and lists of protein IDs as values.

```python
#%%time
translation = {}
i = 1
for gene in loci:
    for transcript in vartranscripts_filtered[gene]:
        if transcript[:3] == 'NM_':
            rec = get_transcript(transcript)
            seq = get_CDS(rec)
            match = False
            for prot in protrecs[gene]: 
                if seq[4] == prot.seq:
                    print(i, gene, transcript, prot.id)
                    match = True
                    translation.update({transcript: prot.id})
            if not match:
                print(i, gene, transcript, 'no matching protein         <<<<<')
        else:
            print(i, gene, transcript, 'invalid transcript name         *****')
        i += 1
```

```python
import os
from Bio import AlignIO

variable = {}
for gene in loci:
    transcripts = vartranscripts_filtered[gene]
    for transcript in transcripts:
        protein = translation[transcript]
        #print(transcript)
        print(protein)

        fasta_file = os.path.join("fasta_match", f"{gene}_{protein}_match.fasta")
        aln_file = os.path.join("aln_match", f"{gene}_{protein}_match.aln")

        #print(protein, fasta_file, aln_file)

        fasta = None
        for record in SeqIO.parse(fasta_file, "fasta"):
            recordid = record.description.split()[0]
            if recordid == protein:
                fasta = record
                break
        if fasta is None:
            raise ValueError(f"{protein} not found in {fasta_file}")    
        #print(fasta)

        align = AlignIO.read(aln_file, "fasta")
        aln = None
        for record in align:
            if record.id == protein:
                aln = record
                break
        if aln is None:
            raise ValueError(f"{protein} not found in {aln_file}")
        #print(aln)
        #print("gene\tHum\tAln\tsite\tvariable_amino_acids")
        
        alni = get_aln_positions(gene, protein)
        L = len(fasta)
        for i in range(L):
            j = alni[i]
            counts = pd.Series(list(align[:, j])).value_counts()
            alleles = counts.index.tolist()
            remove_element(alleles, "-")
            n = len(alleles)
            remove_element(alleles, fasta[i])
            if n > 1:
                variable.setdefault(transcript, {})
                variable[transcript][i + 1] = alleles


                #variable.update({i + 1: alleles})
                #variable.update({transcript: {i + 1: alleles}})
                #variable.update({transcript: alleles})
                #variable[(transcript, i + 1)] = alleles
                #if verbose:
                
                print(gene, transcript, protein, i + 1, j + 1, fasta[i], alleles, sep="\t")

    #variable.update({transcript: get_variable_sites(gene, protein, False)})
```

```python
results = []

for gene in loci:
    transcripts = vartranscripts_filtered[gene]
    for transcript in transcripts:
        protein = translation[transcript]
        fasta_file = os.path.join("fasta_match", f"{gene}_{protein}_match.fasta")
        aln_file = os.path.join("aln_match", f"{gene}_{protein}_match.aln")

        print(f"\n{gene}")
        print(transcript, protein)

        fasta = None

        for record in SeqIO.parse(fasta_file, "fasta"):
            if record.description.split()[0] == protein:
                fasta = record
                break
        if fasta is None:
            raise ValueError(f"{protein} not found")


        align = AlignIO.read(aln_file, "fasta")
        aln = None
        for record in align:
            if record.id == protein:
                aln = record
                break
        if aln is None:
            raise ValueError(f"{protein} not found")


        alni = get_aln_positions(gene, protein)

        for i in range(len(fasta)):
            j = alni[i]
            counts = pd.Series(list(align[:, j])).value_counts()
            alleles = counts.index.tolist()
            remove_element(alleles, "-")
            n = len(alleles)
            remove_element(alleles, fasta.seq[i])

            if n > 1:
                results.append({
                    "gene": gene,
                    "transcript": transcript,
                    "protein": protein,
                    "human_site": i + 1,
                    "alignment_site": j + 1,
                    "reference_aa": str(fasta.seq[i]),
                    "variable_amino_acids": ",".join(alleles)
                })

variable_df = pd.DataFrame(results)
variable_df.to_csv(
    "Variable_sites_match_30JUL.tsv",
    sep="\t",
    index= False)
```

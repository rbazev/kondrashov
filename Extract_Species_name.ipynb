```python
from Bio import AlignIO
from pathlib import Path
import re
import json

# Directory containing alignment files
aln_folder = Path("C:\\Users\\adesi\\OneDrive - University Of Houston\\Documents\\Ricardo_lab\\GitHub\\kondrashov\\aln_lociii_match")

# Store species lists for each protein
species_per_protein = {}

for protein in CPD_file:

    infile = aln_folder / f"{protein}_match.aln"

    if not infile.exists():
        print(f"Alignment not found: {protein}")
        continue

    # Read alignment (FASTA format)
    alignment = AlignIO.read(infile, "fasta")

    species_list = []

    for record in alignment:

        # Extract species name from square brackets
        match = re.search(r"\[([^\]]+)\]", record.description)

        if match:
            species = match.group(1).replace(" ", "_")
        else:
            species = record.id.replace(" ", "_")

        if species not in species_list:
            species_list.append(species)

    # Store species list for each protein
    species_per_protein[protein] = species_list

    print(f"{protein}: {len(species_list)} species")
```

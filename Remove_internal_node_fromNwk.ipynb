```python
"/Users/adesinamary/Library/CloudStorage/OneDrive-UniversityOfHouston/Documents/Ricardo_lab/iqtree-3.1.4-Windows/untitled folder/NWK_FILES"


from pathlib import Path
import re

input_dir = Path("/Users/adesinamary/Library/CloudStorage/OneDrive-UniversityOfHouston/Documents/Ricardo_lab/iqtree-3.1.4-Windows/untitled folder/NWK_FILES")

output_dir = input_dir / "Newick_cleaned"
output_dir.mkdir(exist_ok=True)

# Find all files ending with species.nwk
tree_files = list(input_dir.glob("*species.nwk"))

for file in tree_files:

    with open(file, "r") as f:
        tree = f.read()

    # Remove quoted numeric internal-node labels
    cleaned_tree = re.sub(r"\)'[0-9]+'", ")", tree)

    # Save as .Newick
    output_file = output_dir / file.name.replace(
        "species.nwk", "species_cleaned.Newick"
    )

    with open(output_file, "w") as f:
        f.write(cleaned_tree)

    print(f"Cleaned: {file.name} -> {output_file.name}")

print(f"\nCompleted! Processed {len(tree_files)} files.")
print(f"Output directory: {output_dir}")
```
